```python
# --- GRUPO 1: Manipulación de Tipos y Conversión ---

# 1. print(): Imprime mensajes en la consola
print("¡Hola, bienvenidos a nuestro taller de Python!")

# 2. len(): Muestra la longitud de una cadena o estructura
texto = "Ingeniería"
print("Longitud de la palabra 'Ingeniería':", len(texto))

# 3. type(): Devuelve el tipo de dato de una variable
numero = 42
print("Tipo de dato de 42:", type(numero))

# 4. str(): Convierte un valor a texto (string)
edad = 25
print("Convertir entero a texto:", str(edad))

# 5. int(): Convierte un valor a número entero
valor_decimal = "10"
print("Convertir texto a entero:", int(valor_decimal))
```

## 2. Grupo de Matemáticas y Numéricas

```python
# --- GRUPO 2: Matemáticas y Numéricas ---

# 6. abs(): Devuelve el valor absoluto de un número
print("Valor absoluto de -15:", abs(-15))

# 7. round(): Redondea un número decimal
print("Redondear 3.14159 a 2 decimales:", round(3.14159, 2))

# 8. pow(): Eleva un número a una potencia (base, exponente)
print("2 elevado a la 3:", pow(2, 3))

# 9. sum(): Suma los elementos de una lista
numeros = [10, 20, 30, 40]
print("Suma de la lista [10, 20, 30, 40]:", sum(numeros))

# 10. max(): Devuelve el valor máximo de una lista o conjunto
print("El número mayor entre (5, 9, 2, 14):", max(5, 9, 2, 14))
```

### Continuación - Grupo de Matemáticas y Numéricas

```python
# --- FUNCIONES 11 A 15: Matemáticas Avanzadas y Conversiones ---

# 11. min(): Devuelve el valor mínimo de una lista o secuencia
print("El número menor entre (5, 9, 2, 14):", min(5, 9, 2, 14))

# 12. divmod(): Devuelve una tupla con el cociente y el residuo de una división entera
print("Cociente y residuo de dividir 17 entre 4:", divmod(17, 4))

# 13. bin(): Convierte un número entero en una cadena de texto en formato binario
print("Representación binaria de 10:", bin(10))

# 14. hex(): Convierte un número entero en una cadena de texto en formato hexadecimal
print("Representación hexadecimal de 255:", hex(255))

# 15. float(): Convierte un valor numérico o texto a número decimal (float)
precio_texto = "99.95"
print("Convertir texto a float:", float(precio_texto))
```
